```python
#finding missing values 
# df.info()
# true --> val missing 
df.isnull()
```

```python
#replacing missing age with mean 
df['Age'].fillna(df['Age'].mean(),inplace=True)
print(df.info())
```

```python
#data type conversion
#converting data types 
df['Pclass']=df['Pclass'].astype('category')
df.info()
```

```python
#ploting survived or not survived graph 
sns.countplot(x='Survived', data=df)
plt.show()
```

```python
#ploting sex based survival 
sns.countplot(x='Survived',hue='Sex',data=df)
plt.show()
```

```python
#selecting the numeric data types and then ploting correlation heatmap 
nmdf = df.select_dtypes(include=['int64', 'float64'])

plt.figure(figsize=(10,6))
sns.heatmap(nmdf.corr(), annot=True, cmap='coolwarm')
plt.show()
```
