```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns 
import sklearn as sk
import numpy as np
```

```python
#printing dimensions of the data
df=pd.read_csv('boston_housing.csv')
df.head()
```

```python
df.describe()
```

```python
#displot for detail trend in data or medv vs count 
sns.displot(data=df.medv, kde=True)
```

```python
# checking null values 
df.isnull().sum()
```

```python
sns.boxplot(data=df)
```

```python
# importing required models and scaler to scale the data
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

X=np.asarray(df.drop(['medv'],axis=1))
y=np.asarray(df['medv'])
# train and test data
x_train,x_test,y_train,y_test=train_test_split(X,y,test_size=0.2)
```

```python
# creating linear model from the LinearRegression model of sk learn
model=sk.linear_model.LinearRegression()

x_train=StandardScaler().fit_transform(x_train)
x_test=StandardScaler().fit_transform(x_test)
# training the model 
model.fit(x_train,y_train)
```

```python
# testing the data 
y_pred = model.predict(x_test)
```

```python
# scatter plot to visualise the result
plt.scatter(y_test,y_pred)
```

```python
# calculating the Mean Absolute Error
from sklearn.metrics import mean_absolute_error

print("MSE ",mean_absolute_error(y_test,y_pred))
```
