```python
import nltk, math, re
import sklearn as sk
```

```python
with open('d1.txt','r') as file:
    doc = file.read()
    
```

```python
print(doc)
```

```python
word_tokens = nltk.word_tokenize(doc)
```

```python
print(word_tokens)
```

```python
stopwords = set(nltk.corpus.stopwords.words('english'))
print(stopwords)
```

```python
afterremovingstopwords = [tokens for tokens in word_tokens if tokens not in stopwords]
print(afterremovingstopwords)
```

```python
tagged = nltk.pos_tag(afterremovingstopwords)
```

```python
print(tagged)
```
