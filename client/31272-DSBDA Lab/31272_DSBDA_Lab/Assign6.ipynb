```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

pd.options.display.max_rows=999
```

```python
df=pd.read_csv('IRIS.csv')
print(df.head())
df.describe()
```

```python
# distribution of sepal_length  
sns.boxplot(x='sepal_length',y='species',data=df)

# gave a single outlier for species IRIS-VIRGINICA
```

```python
# bar graph plot to get a relation between sepal and petal growth in different species 
#
sns.barplot(x='sepal_length',y='petal_length',hue='species',data=df)    
```

```python
# count vs sepal length
sns.boxplot(x='species',y='sepal_length',data=df)
plt.show()

# shows sepal length per species 
sns.histplot(x='sepal_length',y='species',data=df)
plt.show()
# tells Iris-setosa's sepal length lies between 2-6
# tells Iris-virginica's sepal length lies between 4.5-8
# tells Iris-versocolor sepal length lies between 4.7-7.2

# and the darkness tells the mode of sepal length for that species
```

```python
sns.boxplot(x='species',y='sepal_width',data=df)
plt.show()
sns.histplot(x='sepal_width',y='species',data=df)
plt.show()
# tells Iris-setosa's sepal width lies between 2.3- 4.5
# tells Iris-versocolor sepal width lies between 2 - 3.5 
# tells Iris-virginica's sepal width lies between 2.5 - 3.5

# and the darkness tells the mode of sepal width for that species
```

```python
sns.boxplot(x='species',y='petal_length',data=df)
plt.show()
sns.histplot(x='petal_length',y='species',data=df)
plt.show()


# tells Iris-setosa's petal length lies between 1-2
# tells Iris-versicolor petal length lies between 3.2-5.2 
# tells Iris-virginica's petal length lies between 4.3-6.9

# and the darkness tells the mode of petal length for that species
```

```python
sns.boxplot(x='species',y='petal_width',data=df)
plt.show()
sns.histplot(x='petal_width',y='species',data=df)
plt.show()


# tells Iris-setosa's petal width lies between 0 - 0.7
# tells Iris-versicolor petal width lies between  1 - 1.75
# tells Iris-virginica's petal width lies between 1.5 - 2.5

# and the darkness tells the mode of petal width for that species
```
