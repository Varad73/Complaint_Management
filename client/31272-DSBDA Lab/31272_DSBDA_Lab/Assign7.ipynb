```python
import nltk ,math,re
import sklearn as sk
nltk.download('punkt')
nltk.download('punkt_tab')
```

## open files

```python
with open('d1.txt','r') as file:
    doc=file.read()

with open('d2.txt','r') as file:
    doc2=file.read()
```

## tokenize the data

```python
word_tokens=nltk.word_tokenize(doc)
print(word_tokens)
word_tokens=nltk.word_tokenize(doc2)
print(word_tokens)
```

## stop words

```python
nltk.download('stopwords')
stop_words = set(nltk.corpus.stopwords.words('english'))
print(stop_words)
```

## remove stop words like the, and, is etc which are of no use 

```python
word_tokens = [token for token in word_tokens if token not in stop_words]
print(word_tokens)
```

## POS tagging :- tagging words by their types or category

```python
# pos tagging
nltk.download('averaged_perceptron_tagger_eng')
tagged = nltk.pos_tag(word_tokens)
print(tagged)
```

## stemming :- cutting down the words to their base words :- removing prefixes or suffixes

```python
stemmer = nltk.stem.PorterStemmer()
stemmed_tokens = [stemmer.stem(token) for token in word_tokens]
print(stemmed_tokens)
```

## lemmatization :- links words with similar meanings to one word.

```python
nltk.download('wordnet')
lemmatizer = nltk.stem.WordNetLemmatizer()
lemmatized_tokens = [lemmatizer.lemmatize(token) for token in word_tokens]
print(lemmatized_tokens)
```

```python

def calc_term_freq(doc):
    word_tokens = nltk.word_tokenize(doc)
    tf_dict = dict()
    for word in word_tokens:
        tf_dict[word] = word_tokens.count(word)
    tf = dict()
    for word, count in tf_dict.items():
        tf[word] = count/len(tf_dict)
    return tf
```

```python

tf = calc_term_freq(doc)
tf2 = calc_term_freq(doc2)

print("Term Frequency of Doc 1: ", tf, "\n\n")
print("Term Frequency of Doc 2: ", tf2)
```

```python
d1_tokens = nltk.word_tokenize(doc)
d2_tokens = nltk.word_tokenize(doc2)

def calculate_idf():
    N = 2
    all_tokens = d1_tokens + d2_tokens
    idf = dict()
    for token in all_tokens:
        f = 0                   # Changed f from 1 back to 0
        if token in d1_tokens:
            f+=1
        if token in d2_tokens:
            f+=1
        idf[token] = math.log(N/f)
    return idf

idf = calculate_idf()
print(idf)
```

```python
d1_repr = []
for token in d1_tokens:
    d1_repr.append(tf[token] * idf[token])
d2_repr = []
for token in d2_tokens:
    d2_repr.append(tf2[token] * idf[token])
```
