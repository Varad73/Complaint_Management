```python
import pandas as pd 
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("titanic.csv")
data.head()
```

```python
sns.barplot(x='Sex', y='Age', data=data)
plt.show()
```

```python
Q1 = data['Age'].quantile(0.25)
Q3 = data['Age'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

# Remove outliers
data = data[(data['Age'] >= lower) & (data['Age'] <= upper)]
```

```python
sns.boxplot(data=data, x="Sex", y="Age", hue="Survived")
```
