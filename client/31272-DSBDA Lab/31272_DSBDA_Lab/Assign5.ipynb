```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

pd.options.display.max_rows=999
```

```python
df=pd.read_csv('Titanic.csv')
df.describe()
```

```python
df.head()
```

```python
# age vs gender vs survived distribution
sns.set_style('dark')
sns.boxplot(x='Sex',y='Age',hue='Survived',data=df)
plt.title('Age Distribution by Gender and Survival')
plt.xlabel('Gender')
plt.ylabel('Age')
```

```python
# observations 
# 1. female had a greater survival rate 
# 2. males had lower survival chances
# 3. for both genders, younger passengers show higher survival
# 4. age spread is wider for males
```
