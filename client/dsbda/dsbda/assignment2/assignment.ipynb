```python
Q1 = df['math score'].quantile(0.25)
Q3 = df['math score'].quantile(0.75)
IQR = Q3 - Q1
lower_limit = Q1 - 1.5 * IQR

upper_limit = Q3 + 1.5 * IQR
outliers = df[
    (df['math score'] < lower_limit) |
    (df['math score'] > upper_limit)
]

print(outliers)
```

```python
# Function to Detect and Handle Outliers using IQR Method

def handle_outliers(col):

    # Step 1 - Calculate Q1 and Q3
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    # Step 2 - Calculate IQR
    IQR = Q3 - Q1

    # Step 3 - Calculate lower and upper limits
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    # Step 4 - Detect outliers
    outliers = df[(df[col] < lower) | (df[col] > upper)]

    # Display outliers
    print("\nOutliers in", col)
    print(outliers[[col]])

    # Step 5 - Handle outliers using clipping
    df[col] = df[col].clip(lower, upper)

    print("Outliers handled successfully for", col)
```

```python
for col in ['math score', 'reading score', 
            'writing score', 'placement score']:

    handle_outliers(col)
```

```python
🔷 METHOD 2 — IQR METHOD

This is the MOST IMPORTANT method for your practical.

Core Idea

Instead of graph,
we use quartiles mathematically.

VERY IMPORTANT IDEA

Outliers are values FAR from middle 50% data.

So:
we first measure middle spread.

Interquartile Range (IQR)

Formula:

IQR=Q3−Q1

Meaning

IQR measures spread of middle 50% data.

Example

Suppose:

Q1	   Q3
60	   80

Then:

IQR=80−60=20

Middle 50% spread = 20.

NEXT IDEA

Now we define:

“How far is TOO FAR?”

Statisticians chose:

1.5×IQR

as reasonable boundary.

Lower Boundary

Q1−1.5(IQR)

Upper Boundary

Q3+1.5(IQR)

Example

Suppose:

Q1	    Q3
60	    80

IQR = 20

Lower Limit
60−1.5(20)=30
Upper Limit
80+1.5(20)=110
Final Rule

Any value:

below 30
above 110

is outlier.

WHY IQR IS POWERFUL

Because it DOES NOT use:

mean
standard deviation

which get affected by extreme values.

Example

Dataset:

Values
70
75
80
500

Mean becomes huge.

But quartiles stay relatively stable.

So IQR still works properly.

WHEN TO PREFER IQR?

✅ Skewed data
✅ Dataset with extreme values
✅ Non-normal distributions
✅ Real-world noisy data

WHY YOUR DATASET USES IQR

Because your dataset already has:

extreme values
skewness

So IQR is best.
```

```python
from scipy import stats
import numpy as np

# Function to Detect Outliers using Z-Score

def detect_zscore_outliers(col):

    # Step 1 - Calculate Z-scores
    z_scores = stats.zscore(df[col])

    # Step 2 - Find outliers
    outliers = df[(z_scores > 3) | (z_scores < -3)]

    # Display outliers
    print("\nOutliers in", col, "using Z-Score Method")
    print(outliers[[col]])
```

```python
for col in ['math score', 'reading score',
            'writing score', 'placement score']:

    detect_zscore_outliers(col)
```

```python
🔷 Main Idea of Z-Score

Z-score answers ONE question:

❓

“How far is a value from the average?”

But not in normal units.

It tells distance in:

👉 standard deviations
First Understand Mean

Suppose marks are:

Marks
60
70
80

Mean:

Mean=(60+70+80)/3 =70

So:
70 is the center of data.

Now Understand Spread

Suppose another dataset:

Marks
69
70
71

Mean is still 70.

But these values are VERY close together.

Another Dataset
Marks
10
70
130

Mean is again 70.

But values are spread far apart.

So Mean Alone is NOT Enough

We also need:

👉 how spread out the data is

This spread is measured using:

🔷 Standard Deviation (SD)
Simple Meaning of SD

Standard deviation tells:

“On average, how far values are from the mean.”

Small SD

Values close to mean.

Example:

Values
69
70
71

SD is small.

Large SD

Values far from mean.

Example:

Values
10
70
130

SD is large.

NOW COMES Z-SCORE 🔥

Z-score combines:

mean
standard deviation

to measure:

“How unusual is a value?”

Formula
Z= (X−μ)/σ
x = 1.2
μ = 0
σ = 1.0
z=(x−μ)/σ≈1.2
Φ(z)≈88.5%
Meaning of Symbols
Symbol	    Meaning
X	        actual value
μ	        mean
σ	        standard deviation
VERY SIMPLE INTERPRETATION
Numerator
X−μ

means:

actual value - average

So it measures distance from mean.

Denominator
σ

standard deviation normalizes the distance.

FINAL MEANING

Z-score tells:

👉

“How many standard deviations away from mean is this value?”

FULL EXAMPLE 🔥

Suppose:

Marks
60
70
80

Mean = 70

SD = 10

Now take:

X=80
Apply Formula

Z=(80−70)/10=1

What Does Z = 1 Mean?

It means:

👉

80 is 1 standard deviation above mean.

Another Example

Take:

X=50
Formula

Z=
(50−70)/10=−2

Meaning

50 is:

2 standard deviations BELOW mean.
IMPORTANT SIGN
Z-score	Meaning
Positive	above mean
Negative	below mean
WHY Z-SCORE DETECTS OUTLIERS

In normal distributions:
most values stay near mean.

Very far values are rare.

General Rule
Z-score	            Meaning
between -1 and 1	very normal
between -2 and 2	mostly normal
beyond ±3	        unusual/outlier
Example

Suppose class marks:

Most Students
60–80

One student gets:

150

Mean = 70
SD = 10

Z-score

Z= (150−70)/10=8

Meaning

150 is:

👉 8 standard deviations away from mean

This is EXTREMELY unusual.

So:
✅ outlier.

WHY Z-SCORE FAILS SOMETIMES

Now VERY IMPORTANT concept.

Suppose:

Values
70
75
80
500
Problem

500 itself changes:

mean
standard deviation
Mean becomes huge

Mean ≠ real center anymore.

SD also becomes huge

Now Z-score calculation becomes distorted.

So:
500 may not appear as extreme as it should.

THAT IS WHY:
Method	       Good For
Z-score	       normal distributions
IQR	           skewed/extreme data
REAL-LIFE ANALOGY 🔥

Suppose average height:

Mean Height
170 cm

SD = 10 cm

Person A = 180 cm

Z-score:

(180−170)/10=1

Normal.

Person B = 250 cm

Z-score:

(250−170)/10=8

Extremely abnormal.

MOST IMPORTANT THING TO REMEMBER

Z-score is NOT measuring:

direct distance

It measures:

👉 relative distance based on spread.
VIVA ANSWERS
Q. What does Z-score represent?

Answer:

Z-score represents how many standard deviations a value lies away from the mean.

Q. Why use ±3 rule?

Answer:

In normal distributions, almost all values lie within 3 standard deviations from the mean. Values beyond this range are considered unusual.

Q. Why is Z-score sensitive to outliers?

Answer:

Because outliers affect both mean and standard deviation, which are used in Z-score calculation.
```
