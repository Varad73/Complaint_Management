```python
def IQR_outlier(col):
    Q1=df[col].quantile(0.25)
    Q2=df[col].quantile(0.75)

    IQR = Q2 - Q1

    lower=Q1-1.5*IQR
    upper=Q2+1.5*IQR

    outliers=df[(lower >df[col]) | (upper<df[col])]

    print("Outliers are : ")
    print(outliers[[col]])

    df[col]=df[col].clip(lower,upper)    

    print("outliers are rmoved")
```

```python
for col in ['math score','reading score','writing score','placement score']:
    IQR_outlier(col)
```
