```python
import pandas as pd
```

```python
# Load dataset
df = pd.read_csv("nba.csv")

# Display first 5 rows
print(df.head())
```

```python
df.isnull().sum()
```

```python
df.shape
```

```python
df = df.dropna(subset=["Team", "Age", "Weight", "Salary"])
```

```python
df.shape
```

```python
result = df.groupby("Team")[["Age", "Weight", "Salary"]].agg(
    ["mean", "median", "min", "max", "std"]
)
```

```python
print("\nSummary Statistics Grouped By Team:\n")
print(result)
```

```python
df = pd.read_csv("Iris.csv")
```

```python
print(df.head())
```

```python
print(df.columns)
```

```python
print(df.isnull().sum())
```

```python
# Statistical summary grouped by species
result = df.groupby("Species")[
    ["SepalLengthCm", "SepalWidthCm",
     "PetalLengthCm", "PetalWidthCm"]
].agg(["mean", "std", "min", "max"])
```

```python

print("\nStatistical Summary:\n")
print(result)
```

```python
# Percentiles
# percentiles = df.groupby("Species").quantile([0.25, 0.5, 0.75])

percentiles = df.groupby("Species")[
    ["SepalLengthCm", "SepalWidthCm",
     "PetalLengthCm", "PetalWidthCm"]
].quantile([0.25, 0.5, 0.75])

print("\nPercentiles:\n")
print(percentiles)
```

```python
v = df[[
    "SepalLengthCm",
    "SepalWidthCm",
    "PetalLengthCm",
    "PetalWidthCm"
]].quantile([0.25])

print(v)
```

```python
# Mode for each species

mode_values = df.groupby("Species")[[
    "SepalLengthCm",
    "SepalWidthCm",
    "PetalLengthCm",
    "PetalWidthCm"
]].agg(lambda x: x.mode()[0])

print("\nMode Values:\n")
print(mode_values)
```
