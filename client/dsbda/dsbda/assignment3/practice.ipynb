```python
import pandas as pd
```

```python
df = pd.read_csv("nba.csv")
```

```python
df.head()
```

```python
df.dtypes
```

```python
df.isnull().sum()
```

```python
df=df.dropna(subset=["Team","Age","Salary","Weight"])
```

```python
df.shape
```

```python
result=df.groupby("Team")[["Age","Salary","Weight"]].agg(["mean","median","min","max","std"])
print(result)
```

```python
df=pd.read_csv("iris.csv")
```

```python
df.dtypes
```

```python
result=df.groupby("Species")[["SepalLengthCm","SepalWidthCm","PetalLengthCm","PetalWidthCm"]].agg(["mean","median","min","max","std"])
print(result)
```

```python
result=df.groupby("Species")[["SepalLengthCm","SepalWidthCm","PetalLengthCm","PetalWidthCm"]].quantile([0.25,0.50,0.75])
```

```python
print(result)
```
