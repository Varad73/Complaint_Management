```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python

# load iris dataset
df = pd.read_csv('Iris.csv')

print("First 5 rows:")
print(df.head())

```

```python
print("\nShape:", df.shape)
```

```python
# Task 1 - List features and their types
print("Features and their Types:")
print(df.dtypes)

# float64 = Numeric
# object  = Nominal (categorical)
```

```python
df.columns
```

```python
df.Species.unique()
```

```python
cols = ["Species", "Id", "SepalLengthCm", "SepalWidthCm", "PetalLengthCm", "PetalWidthCm"]

for col in cols:
    if col == "Species" or col == "Id":  # ✅ compare col not df[col]
        continue
    
    sns.histplot(data=df, x=col,bins=20,kde=True)   # ✅
    plt.title(f"Histogram of {col}")
    plt.show()
```

```python
# Task 3 - Boxplot for each feature
for column in df.columns:
    
    # skip Id and Species column
    if df[column].dtype == 'object' or column == 'Id':
        continue
    
    plt.figure(figsize=(6, 4))
    sns.boxplot(data=df, y=column)
    plt.title(f"Boxplot of {column}")
    plt.ylabel(column)
    plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.boxplot(data=df[['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm']])
plt.title(f"Boxplot of all columns")
plt.ylabel(column)
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.histplot(data=df[['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm']],bins=20)
plt.title(f"Boxplot of all columns")
plt.ylabel(column)
plt.show()
```
