```python
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("Iris.csv")
```

```python
df.head()
```

```python
df.columns
```

```python
df.Species.unique()
```

```python
cols=["SepalLengthCm","SepalWidthCm","PetalLengthCm","PetalWidthCm"]
```

```python
for col in cols:
    plt.figure(figsize=(8,6))
    sns.histplot(data=df,x=df[col],bins=30,kde=True)
    plt.title(f"Histogram for {col} ")
    plt.xlabel({col})
    plt.ylabel("count")
    plt.show()
```

```python
for col in cols:
    plt.figure(figsize=(8,6))
    sns.boxplot(data=df,y=df[col])
    plt.title(f"Boxplot for {col} ")
    plt.xlabel({col})
    plt.ylabel("count")
    plt.show()
```
