```python
with open('document1.txt','r') as file:
    text=file.read()
print(text)
```

```python
tokens=word_tokenize(text)
print(tokens)
```

```python
pos=pos_tag(tokens)
print(pos)
```

```python
stop_words=stopwords.words('english')
filtered=[]
for word in tokens:
    if word.lower() not in stop_words:
        filtered.append(word)
print(filtered)
```

```python
stemmer=PorterStemmer()
s=[]
for word in filtered:
    s.append(stemmer.stem(word))
print(s)
```

```python
lemmatizer=WordNetLemmatizer()
l=[]
for word in filtered:
    l.append(lemmatizer.lemmatize(word))
print(l)
```

```python
with open('document1.txt','r') as file:
    doc1=file.read()
print('First Document ')
print(doc1)
    
```

```python
with open('document2.txt','r') as file:
    doc2=file.read()
print('First Document ')
print(doc2)
    
```

```python
#TF
def TF(doc):
    word=doc.lower().split()
    N=len(word)
    tf={}

    word_count=Counter(word)

    for w,c in word_count.items():
        tf[w]=c/N

    return tf
    
```

```python
tf1=TF(doc1)
print(tf1)
tf2=TF(doc2)
print(tf2)
```

```python
#TF
def IDF(docs):
    N=len(docs)
    idf={}
    idf2={}
    
    for doc in docs:
        unique_word=set(doc.lower().split())

        for word in unique_word:
            if word in idf:
                idf[word]+=1
            else:
                idf[word]=1

    for word,count in idf.items():
        idf2[word]=math.log(N/idf[word])

    return idf2;
```

```python
doc=[doc1,doc2]
i=IDF(doc)
print(i)
```

```python
#TF-IDF
itidf1={}
for word,count in tf1.items():
    itidf1[word]=count*i[word]

print(itidf1)


itidf2={}
for word,count in tf2.items():
    itidf2[word]=count*i[word]

print(itidf2)


    
```
