```python
# NLP Library
import nltk

# Mathematical operations
import math

# Used for counting word frequency
from collections import Counter

# Tokenization
from nltk.tokenize import word_tokenize,sent_tokenize

# Stopword removal
from nltk.corpus import stopwords

# Stemming
from nltk.stem import PorterStemmer

# Lemmatization
from nltk.stem import WordNetLemmatizer

# POS Tagging
from nltk import pos_tag
```

```python
with open('document1.txt', 'r') as file:
    
    text = file.read()

print("Original Document:\n")

print(text)
```

```python
tokens = word_tokenize(text)

print("Tokens:\n")

print(tokens)
```

```python
sentances = sent_tokenize(text)

print("Tokens:\n")

print(sentances)
```

```python
pos = pos_tag(tokens)

print("POS Tags:\n")

print(pos)
```

```python
# Get English stopwords
stop_words = set(stopwords.words('english'))

# Empty list for filtered words
filtered = []

# Check every token
for word in tokens:
    
    # Convert to lowercase and check
    if word.lower() not in stop_words:
        
        filtered.append(word)

print("After Stopword Removal:\n")

print(filtered)
```

```python
# Create stemmer object
stemmer = PorterStemmer()

# Empty list
stemmed = []

# Apply stemming on each word
for word in filtered:
    
    stemmed.append(stemmer.stem(word))

print("Stemmed Words:\n")

print(stemmed)
```

```python
# Create lemmatizer object
lemmatizer = WordNetLemmatizer()

# Empty list
lemmatized = []

# Apply lemmatization
for word in filtered:
    
    lemmatized.append(lemmatizer.lemmatize(word))

print("Lemmatized Words:\n")

print(lemmatized)
```

```python
# Read first document
with open('document1.txt', 'r') as file:
    
    doc1 = file.read()
  
# Read second document
with open('document2.txt', 'r') as file:
    
    doc2 = file.read()

# Store both documents in list
docs = [doc1, doc2]

print(docs)
```

```python
from collections import Counter

# Function to calculate TF
def calculate_tf(doc):

    # Convert document into words
    words = doc.lower().split()

    # Total number of words
    total_words = len(words)

    # Count frequency of each word
    word_count = Counter(words)

    # Empty dictionary
    tf = {}

    # Calculate TF
    for word, count in word_count.items():

        tf[word] = count / total_words

    return tf

# Calculate TF for document 1
tf1 = calculate_tf(doc1)

print("TF of Document 1:\n")

print(tf1)
```

```python
# Total number of documents
N = len(docs)

# Dictionary to store document count
doc_count = {}

# Check each document
for doc in docs:

    # Get unique words from document
    unique_words = set(doc.lower().split())

    # Count in how many documents word appears
    for word in unique_words:

        if word in doc_count:

            doc_count[word] += 1

        else:

            doc_count[word] = 1

# Empty dictionary for IDF
idf = {}

# Calculate IDF
for word, count in doc_count.items():

    idf[word] = math.log(N / count)

print("IDF Values:\n")

print(idf)
```

```python
# TF
tf1 = calculate_tf(doc1)
tf2 = calculate_tf(doc2)

# TF-IDF for doc1
tfidf1 = {}

for word in tf1:

    tfidf1[word] = tf1[word] * idf[word]

# TF-IDF for doc2
tfidf2 = {}

for word in tf2:

    tfidf2[word] = tf2[word] * idf[word]
```

```python
# TF-IDF for doc1
print(tfidf1)
```

```python
print(tfidf2)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

import pandas as pd

# Store both documents
documents = [doc1, doc2]

# Create vectorizer object
vectorizer = TfidfVectorizer()

# Generate TF-IDF matrix
tfidf_matrix = vectorizer.fit_transform(documents)

# Get all unique words
feature_names = vectorizer.get_feature_names_out()

# Convert matrix into dataframe
df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=feature_names
)

print(df)
```
