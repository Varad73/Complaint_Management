```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("Titanic-Dataset.csv")
```

```python
print(df.head())
```

```python
print(df.info())
```

```python

print(df.describe())
```

```python
df.isnull().sum()
```

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
```

```python
df["Fare"] = df["Fare"].fillna(df["Fare"].median())
```

```python
plt.figure(figsize=(8,6))

sns.histplot(data=df, x="Fare", bins=30)

plt.title("Histogram of Ticket Fare")

plt.xlabel("Fare")

plt.ylabel("Number of Passengers")

plt.show()
```

```python
plt.figure(figsize=(6,5))

sns.countplot(data=df, x="Survived")

plt.title("Count of Survived Passengers")

plt.xlabel("Survived")

plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.countplot(data=df, x="Sex", hue="Survived")

plt.title("Survival Count Based on Gender")

plt.xlabel("Gender")

plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.countplot(data=df, x="Survived", hue="Sex")

plt.title("Survival Count Based on Gender")

plt.xlabel("Gender")

plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(6,5))

sns.countplot(data=df, x="Sex")

plt.title("Count of Male and Female Passengers")

plt.xlabel("Gender")

plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.boxplot(x="Pclass", y="Age", data=df)

plt.title("Age Distribution by Passenger Class")

plt.xlabel("Passenger Class")

plt.ylabel("Age")

plt.show()
```

```python
plt.figure(figsize=(6,8))

sns.boxplot(data=df, y="Fare")

plt.title("Vertical Boxplot of Fare")

plt.ylabel("Fare")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.barplot(data=df, x="Pclass", y="Fare")
plt.title("Average Fare by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Average Fare")
plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(data=df, x="Age", y="Fare")

plt.title("Scatterplot of Age vs Fare")

plt.xlabel("Age")

plt.ylabel("Fare")

plt.show()
```

```python
plt.figure(figsize=(10,8))

correlation = df.corr(numeric_only=True)

sns.heatmap(correlation, annot=True)

plt.title("Correlation Heatmap")

plt.show()
```

```python

df["Age"] = df["Age"].fillna(df["Age"].median())

# Plot the histogram
plt.figure(figsize=(8, 6))
sns.histplot(data=df, x="Age", bins=20, kde=False)
plt.title("Histogram of Passenger Age")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()
```
