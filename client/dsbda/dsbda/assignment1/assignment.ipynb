```python
HOW TO CHECK MISSING VALUES
df.isnull().sum()

This tells how many missing values each column has.

DIFFERENT WAYS TO HANDLE MISSING VALUES
1. Fill with Mean (MOST COMMON)

Used for:

✅ Numerical continuous data

Examples:

Age
Salary
Height
Weight
Code
df['Age'] = df['Age'].fillna(df['Age'].mean())
Why Mean?

Because it keeps average distribution balanced.

When NOT to use Mean

❌ If data has many outliers.

Example:

10, 12, 15, 1000

Mean becomes very large.

2. Fill with Median

Used when:

✅ Numerical data has outliers

Examples:

Income
House price
Fare
Code
df['Fare'] = df['Fare'].fillna(df['Fare'].median())
Why Median?

Median is not affected by outliers.

Example

Data:

10, 12, 15, 1000

Mean = huge

Median = 13.5 (better)

3. Fill with Mode

Used for:

✅ Categorical/Text data

Examples:

Gender
City
Embarked
Code
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
Why Mode?

Mode = most frequent value.

Example
Gender
Male
Female
Male
NaN

Fill with:

Male

because it appears most.

4. Drop Rows

Used when:

✅ Very few missing rows

Code
df.dropna(inplace=True)
Why?

If only few rows missing, removing them is okay.

When NOT to Use

❌ If dataset becomes too small.

5. Drop Columns

Used when:

✅ Column has TOO MANY missing values.

Example:

Cabin
NaN
NaN
NaN
Code
df.drop('Cabin', axis=1, inplace=True)
Why?

Column has little useful information.

TITANIC DATASET EXAMPLE
Column	       Best Method	    Why
Age	           Mean	            Numerical continuous
Fare	       Median	        Possible outliers
Embarked	   Mode	            Categorical
Cabin	       Drop column	    Too many missing values
    
HOW TO DECIDE WHAT TO USE
Rule of Thumb
    
Data Type	                                        Method
Numerical normal data	        -->                  Mean
Numerical with outliers	        -->                  Median
Categorical/Text	            -->                  Mode
Too many missing rows	        -->                  Drop rows
Too many missing column values	-->                  Drop column
    
IMPORTANT VIVA QUESTION
Why not always drop missing values?

Answer:
Dropping data may reduce dataset size and important information may be lost.

Another Important Viva
Difference between Mean and Median?
Mean	                   Median
Average value	           Middle value
Affected by outliers	   Not affected
BEST ANSWER FOR YOUR TITANIC ASSIGNMENT

You can say:

Age column missing values are handled using mean because Age is numerical continuous data.

Cabin column can be dropped because it contains many missing values.

Embarked missing values can be handled using mode because it is categorical data.


    df['Fare'] = df['Fare'].fillna(df['Fare'].median())
this just a example 
```
